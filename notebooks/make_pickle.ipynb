# Save iJL1678b-ME as a pickle

In case the available iJL1678b-ME.pickle file breaks, create a new pickle of the JSON file

```python
import pickle
from zipfile import ZipFile
from cobrame.io.json import load_json_me_model
```

```python
from pathlib import Path
import sys
import os

ROOT = Path(os.getcwd())

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

FILES = ROOT / "files"

sys.path.append(str(ROOT))
```

1. Unzip:

```python
zip_path = FILES / "iJL1678b.zip"

with ZipFile(zip_path, "r") as zipf:
    zipf.extractall(FILES)
```

2. Load as JSON file:

```python
model = load_json_me_model(FILES/"iJL1678b.json")
```

3. Save as picke:

```python
with open(FILES/'iJL1678b.pickle', 'wb') as f:
    pickle.dump(model, f)
```
