```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
lo_samples = pd.read_pickle('/triumvirate/home/celinaha/ml_pipeline/lo_samples.pkl')
```

```python
lo_plot = lo_samples.drop(columns=['sample','biomass_dilution', 'source'])
lo_plot.columns = lo_plot.columns.str.split('_').str[0]
lo_plot = lo_plot.rename(columns={'dummy': 'dummy protein'})
crest = sns.color_palette("crest", as_cmap=True)


plt.figure(figsize=(10,6))
sns.boxplot(data=lo_plot, log_scale=True, color=crest(0.20))
plt.xlabel('Biomass Component')
plt.ylabel("Flux")
plt.xticks(rotation=90) 
plt.title('LO-samples')
plt.show()
```

```python
lo_plot = lo_samples.drop(columns=['sample','biomass_dilution', 'source'])
lo_plot.columns = lo_plot.columns.str.split('_').str[0]
lo_plot = lo_plot.div(lo_samples['biomass_dilution'], axis=0)
lo_plot = lo_plot.rename(columns={'dummy': 'dummy protein'})
crest = sns.color_palette("crest", as_cmap=True)


plt.figure(figsize=(10,6))
sns.boxplot(data=lo_plot, log_scale=True, color=crest(0.20))
plt.xlabel('Biomass Component')
plt.ylabel("Flux")
plt.xticks(rotation=90)
plt.title('LO-samples (growth normalized)')
plt.show()
```
